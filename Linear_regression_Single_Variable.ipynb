```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = 'https://raw.githubusercontent.com/shahjamanbadhon/Data-Sets/refs/heads/main/nasdaq100.csv'
```

```python
df = pd.read_csv(url, sep=';')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
#df.drop('Date', axis=1)
df = df.drop(columns = ['Date'])
```

```python
df.head()
```

```python
plt.scatter(df['Starting (USD)'], df['Ending (USD)'])
plt.xlabel("Starting (USD)")
plt.ylabel("Ending (USD)")
plt.title('NASDAQ100 Stock Prices')
plt.show()
```

```python
x = df.drop('Ending (USD)', axis = 1)
x.head()
```

```python
y = df[['Ending (USD)']]
y.head()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
x.mean()
```

```python
y.mean()
```

```python
plt.scatter(x.mean(),y.mean(), color = 'red')
plt.scatter(df['Starting (USD)'], df['Ending (USD)'])
plt.xlabel("Starting (USD)")
plt.ylabel("Ending (USD)")
plt.title('NASDAQ100 Stock Prices')
plt.show()
```

```python
reg.fit(x,y)
```

```python
m = reg.coef_
```

```python
c = reg.intercept_
```

```python
# y = mx + c
m * 16700 + c
```

```python
reg.predict([[16700]])
```

```python
df['predicted_y'] = reg.predict(x)
df.head()
```

```python
plt.plot(x,df['predicted_y'])
#plt.plot(x, reg.predict(x))
plt.scatter(x, reg.predict(x))

plt.scatter(x.mean(),y.mean(), color = 'red')
plt.scatter(df['Starting (USD)'], df['Ending (USD)'])
plt.xlabel("Starting (USD)")
plt.ylabel("Ending (USD)")
plt.title('NASDAQ100 Stock Prices')
plt.show()
```
